```python
# 1- import the random module
import random

# 2 - crate subject
subjects = [
    "Aojo deb",
    "jubayer",
    "jihan",
    "Sayem",
    "Johura",
    "He",
    "She"
]

actions = [
    "launches",
    "cancels",
    "eats ",
    "orders",
    "goes",
    "come",
    "make",
    "dances with",
    "declares war on",
    "celebrates",
    "orders"
]

places_or_things = [
    "at Red Fort",
    "in Mubai local train",
    "inside parliament",
    " a pices pizza",
    "a plate of samosa",
    "inside ",
    "during IPL match".
    "In USA"
]

# subject action places| objecets
# 3 start the heading generation loop
while True:
    subject =random.choice(subjects)
    action = random.choice(actions)
    place_or= random.choice(places_or_things)

    headline = f" BREAKING NEWS : {subject} {action} {place_or}"

    print("\n" + headline)

    user_input = input("\n Do you want another headlie ?(Y/N)").strip().lower()

    if user_input == "n":
        break


# print goodbye message
print("\n Thanks for using the fake news headline")
```
